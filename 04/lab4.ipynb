```python
import os
import random
import time

import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim


import gym
from datetime import datetime
```

```python

# Imports all our hyperparameters from the other file
from hyperparams import Hyperparameters as params

# print all attributes and their values
for attr in dir(params):
    if not attr.startswith("__"):
        print(f"{attr} = {getattr(params, attr)}")
# stable_baselines3 have wrappers that simplifies 
# the preprocessing a lot, read more about them here:
# https://stable-baselines3.readthedocs.io/en/master/common/atari_wrappers.html
from stable_baselines3.common.atari_wrappers import (
    ClipRewardEnv,
    EpisodicLifeEnv,
    FireResetEnv,
    MaxAndSkipEnv,
    NoopResetEnv,
)
from stable_baselines3.common.buffers import ReplayBuffer


# Creates our gym environment and with all our wrappers.
def make_env(env_id, seed, idx, capture_video, run_name):
    def thunk():
        env = gym.make(env_id)
        env = gym.wrappers.RecordEpisodeStatistics(env)
        if capture_video:
            if idx == 0:
                env = gym.wrappers.RecordVideo(env, f"videos/{run_name}")
        env = NoopResetEnv(env, noop_max=30)
        env = MaxAndSkipEnv(env, skip=4)
        env = EpisodicLifeEnv(env)
        if "FIRE" in env.unwrapped.get_action_meanings():
            env = FireResetEnv(env)
        env = ClipRewardEnv(env)
        env = gym.wrappers.ResizeObservation(env, (84, 84))
        env = gym.wrappers.GrayScaleObservation(env)
        env = gym.wrappers.FrameStack(env, 4)
        env.seed(seed)
        env.action_space.seed(seed)
        env.observation_space.seed(seed)
        return env

    return thunk


# class QNetworkOld(nn.Module):
#     def __init__(self, env):
#         super().__init__()
#         # TODO: Deinfe your network (agent)
#         # Look at Section 4.1 in the paper for help: https://arxiv.org/pdf/1312.5602v1.pdf
#         self.network = nn.Sequential(
#             nn.Conv2d(4, 32, 8, stride=4),
#             nn.ReLU(),
#             nn.Conv2d(32, 64, 4, stride=2),
#             nn.ReLU(),
#             nn.Conv2d(64, 64, 3, stride=1),
#             nn.ReLU(),
#             nn.Flatten(),
#             nn.Linear(3136, 512),
#             nn.ReLU(),
#             nn.Linear(512, env.single_action_space.n)                
#         )

#     def forward(self, x):
#         return self.network(x / 255.0)


class QNetwork(nn.Module):
    def __init__(self, env):
        super().__init__()
        # TODO: Deinfe your network (agent)
        # Look at Section 4.1 in the paper for help: https://arxiv.org/pdf/1312.5602v1.pdf
        # Input to the 
        self.network = nn.Sequential(
            # 2D Convolutional layer
            # 4 Input Channels representing the 4 stacked frames
            # 16 Output channels => the number of filters/kernels for this layer, 16 feature maps will be produced
            # These represents learned spatial features from the input data
            # Kernel_size = 8 => 8x8
            # Stride = 4 => Filter movement for each step more or less a downsampling
            # Output size = (Input_size - kernel_size) / Stride +1
            # (84-8)/4+1=20 (same for both height and width)
            # Input size, output size, kernel size, stride
            nn.Conv2d(4, 16, kernel_size=8, stride=4),  # Output: [batch, 16, 20, 20] 84x84x4 → 20x20x16
            
            nn.ReLU(),
            
            # 16 Feature maps of 20x20 as input, 32 maps of 9x9 as output
            nn.Conv2d(16, 32, kernel_size=4, stride=2),  # Output: [batch, 32, 9, 9] 20x20x16 → 9x9x32
            
            nn.ReLU(),
            
            # Flatten into 2d tensor
            nn.Flatten(),  # Output: [batch, 2592] 32*9*9 = 2592
            
            # 2592 input features => 256 output features
            nn.Linear(2592, 256),
            
            nn.ReLU(),

            # Maps the 256 input features to the number of actions in the game environment
            nn.Linear(256, env.single_action_space.n)                
        )

    def forward(self, x):
        # Normalize input image
        return self.network(x / 255.0)


def linear_schedule(start_e: float, end_e: float, duration: int, t: int):
    slope = (end_e - start_e) / duration
    return max(slope * t + start_e, end_e)

if __name__ == "__main__":
    run_name = f"{params.env_id}__{params.exp_name}__{params.seed}__{int(time.time())}"

    random.seed(params.seed)
    np.random.seed(params.seed)
    torch.manual_seed(params.seed)
    torch.backends.cudnn.deterministic = params.torch_deterministic

    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

    # env setup
    envs = gym.vector.SyncVectorEnv([make_env(params.env_id, params.seed, 0, params.capture_video, run_name)])
    assert isinstance(envs.single_action_space, gym.spaces.Discrete), "only discrete action space is supported"


    # print(envs)
    for attr in dir(envs):
        if not attr.startswith("__"):
            print(f"{attr} = {getattr(envs, attr)}")
    
    q_network = QNetwork(envs).to(device)
    optimizer = optim.Adam(q_network.parameters(), lr=params.learning_rate)
    # Target network to be used for stability 
    target_network = QNetwork(envs).to(device)
    target_network.load_state_dict(q_network.state_dict())
    
    # We’ll be using experience replay memory for training our DQN. 
    # It stores the transitions that the agent observes, allowing us to reuse this data later. 
    # By sampling from it randomly, the transitions that build up a batch are decorrelated. 
    # It has been shown that this greatly stabilizes and improves the DQN training procedure.
    rb = ReplayBuffer(
        params.buffer_size,
        envs.single_observation_space,
        envs.single_action_space,
        device,
        optimize_memory_usage=False,
        handle_timeout_termination=True,
    )

    # Flag to control if training should resume or not
    resume_training = True

    # Update with the run path for resuming model training state
    model_path = f"runs/BreakoutNoFrameskip-v4__DQN_Breakout__1__1748107210/{params.exp_name}_model"
    
    if resume_training:
        # Load the checkpoint
        checkpoint = torch.load(model_path)
        
        # Restore model and optimizer state
        q_network.load_state_dict(checkpoint["model_state_dict"])
        optimizer.load_state_dict(checkpoint["optimizer_state_dict"])
        target_network.load_state_dict(q_network.state_dict())
        start_step = checkpoint["step"] + 1
    
    prev_global_step = 0

    # How often to write the state out to disk
    save_interval = 10000
    
    # Keep track if we have reached the number of cycles prior to training
    training_mode = False
    
    obs = envs.reset()
    
    for global_step in range(start_step, params.total_timesteps):
        # Here we get epsilon for our epislon greedy.
        # linear decay schedule to move from exploration to exploitation
        epsilon = linear_schedule(params.start_e, params.end_e, params.exploration_fraction * params.total_timesteps, global_step)

        if random.random() < epsilon:
            actions = np.array([envs.single_action_space.sample()]) # TODO: sample a random action from the environment 
        else:
            # q_values = # TODO: get q_values from the network you defined, what should the network receive as input?
            obs_tensor = torch.tensor(obs, device=device).float()
            q_values = q_network(obs_tensor)
            actions = torch.argmax(q_values, dim=1).cpu().numpy()

        # Take a step in the environment
        next_obs, rewards, dones, infos = envs.step(actions)

        # Here we print our reward.
        for info in infos:
            if "episode" in info.keys():
                now = datetime.now()
                print(f"global_step={global_step}, episodic_return={info['episode']['r']} @ {now:%Y-%m-%d %H:%M:%S}")
                break

        # Save data to replay buffer
        real_next_obs = next_obs.copy()
        for idx, d in enumerate(dones):
            if d:
                real_next_obs[idx] = infos[idx]["terminal_observation"]

        # Here we store the transitions in D
        rb.add(obs, real_next_obs, actions, rewards, dones, infos)

        obs = next_obs
        # Training 
        if global_step > params.learning_starts:
            if training_mode == False:
                if resume_training:
                    print("....Training Resumes")
                else:
                    print("....Training Starts")
                training_mode = True
            if global_step % params.train_frequency == 0:
                # Sample random minibatch of transitions from D
                data = rb.sample(params.batch_size)
                # You can get data with:
                # data.observations, data.rewards, data.dones, data.actions

                with torch.no_grad():
                    next_q_values = target_network(data.next_observations)
                    # Now we calculate the y_j for non-terminal phi.
                    # Find best possible action
                    target_max, _ = torch.max(next_q_values, dim=1) # TODO: Calculate max Q
                    td_target = data.rewards.flatten() + (1 - data.dones.flatten()) * params.gamma * target_max # TODO: Calculate the td_target (y_j)

                # old_val = q_network(?).gather(1, data.actions).squeeze()
                # loss = F.mse_loss(?, ?) 
                old_val = q_network(data.observations).gather(1, data.actions).squeeze()
                loss = F.mse_loss(old_val, td_target)

                # perform our gradient decent step
                optimizer.zero_grad()
                loss.backward()
                optimizer.step()

                # Save training progress to be able to resume
                if (global_step - prev_global_step) > save_interval:
                    model_path = f"runs/{run_name}/{params.exp_name}_model"
                    os.makedirs(f"runs/{run_name}", exist_ok=True)
                    
                    torch.save({
                        "step": global_step,
                        "model_state_dict": q_network.state_dict(),
                        "optimizer_state_dict": optimizer.state_dict()
                    }, model_path)
                    prev_global_step = global_step
                    print(f"model saved to {model_path}, next save is at > {global_step+save_interval}")
                    
            # update target network
            if global_step % params.target_network_frequency == 0:
                for target_network_param, q_network_param in zip(target_network.parameters(), q_network.parameters()):
                    target_network_param.data.copy_(
                        params.tau * q_network_param.data + (1.0 - params.tau) * target_network_param.data
                    )

    if params.save_model:
        model_path = f"runs/{run_name}/{params.exp_name}_model"
        torch.save(q_network.state_dict(), model_path)
        print(f"model saved to {model_path}")

    envs.close()
```
